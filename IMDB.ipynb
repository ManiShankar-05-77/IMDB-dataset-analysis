* ## **Identifying and handeling missing values**

```python
# finding missing data in columnns 
df.isnull().sum()
# 128 missing valuse in 'revenue(millions)'. 
# 64 missing values in 'metascore'.
```

```python
# filling the missing values in both columns by median
df[["Metascore",'Revenue (Millions)']] = df[["Metascore","Revenue (Millions)"]].fillna(df[["Metascore","Revenue (Millions)"]].median())
```

```python
# Checking if 'year' column contain values stored as strings
df["Year"].isnull().sum()
```

```python
# Checking if 'Runtime (Minutes)' column contain values stored as strings
df["Runtime (Minutes)"].isnull().sum()
```

* ## **Exploratory Data analysis** 

 ### 1. **Top 10 Highest Rated Movies**

```python
top_rated = df.sort_values(by="Rating",ascending=False).head(10)

print(top_rated[["Title", "Rating", "Year"]])
```

 ### 2. **Top 10 movies directors**

```python
top_rated = df.sort_values(by="Rating",ascending=False).head(10)

print(top_rated[["Title","Director"]])
```

 ### 3. **Top 10 movies based on the revenue**

```python
top_rated = df.sort_values(by="Revenue (Millions)",ascending=False).head(10)

print(top_rated[["Title","Revenue (Millions)"]])
```

 ### 4. **visualization of Distribution of IMDb Ratings**

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    df["Rating"].dropna(),
    bins=10,
    kde=True
)

plt.title("Distribution of IMDb Ratings")
plt.xlabel("IMDb Rating")
plt.ylabel("Number of Movies")

plt.show()
```

 ### 5. **visualization of Number of Movies Released by Year**

```python
movies_per_year = df.groupby("Year").size()

plt.figure(figsize=(12, 6))

movies_per_year.plot(marker='o')
plt.title("Number of Movies Released by Year")
plt.xlabel("Year")
plt.ylabel("Number of Movies")
plt.grid(True)
plt.show()
```

 ### 6. **Top 10 movies by highest runtime**

```python
top_rated = df.sort_values(
    by="Runtime (Minutes)",
    ascending=False
).head(10)

print(top_rated[
    ["Title","Runtime (Minutes)","Rating","Revenue (Millions)"]
])
```

 ### 7. **Top 10 movies by lowest runtime**

```python
top_rated = df.sort_values(
    by="Runtime (Minutes)",
    ascending=True
).head(10)

print(top_rated[
    ["Title","Runtime (Minutes)","Rating","Revenue (Millions)"]
])
```

 ### 8. **Top 10 Directors by Number of Movies and its visualization**

```python
director_counts = df["Director"].value_counts()

print(director_counts.head(10))
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=director_counts.head(10).values,
    y=director_counts.head(10).index
)

plt.title("Top 10 Directors by Number of Movies")
plt.xlabel("Number of Movies")
plt.ylabel("Director")

plt.show()
```

 ### 9. **Average rating of movies of Directors**

```python
director_rating = df.groupby(
    "Director"
)["Rating"].agg(
    ["mean", "count"]
)
director_rating = director_rating[
    director_rating["count"] >= 3
]
director_rating = director_rating.sort_values(
    by="mean",
    ascending=False
)
print(director_rating.head(10))
```

 ### 10. **Most voted movies with rating and its visualization**

```python
most_voted = df.sort_values(
    by="Votes",
    ascending=False
).head(10)
print(
    most_voted[
        ["Title", "Votes", "Rating"]
    ]
)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=most_voted,
    x="Votes",
    y="Title"
)

plt.title("Top 10 Movies by Number of Votes")
plt.xlabel("Number of Votes")
plt.ylabel("Movie")

plt.show()
```

 ### 11. **Categorizing movies as Excellent,Good and Average and showing it in chart**

```python
def rating(rating):
    if rating>=7.0:
        return 'Excellent'
    elif rating>=6.0:
        return 'Good'
    else:
        return 'Average'

df['category']=df['Rating'].apply(rating)       
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="category",
    order=["Average", "Good", "Excellent"]
)

plt.title("IMDb Movies by Rating Category")
plt.xlabel("Movie Category")
plt.ylabel("Number of Movies")

plt.show()
```

 ### 12. **Movie Runtime vs IMDb Rating**

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Runtime (Minutes)",
    y="Rating",
)

plt.title("Movie Runtime vs IMDb Rating")
plt.xlabel("Runtime (Minutes)")
plt.ylabel("IMDb Rating")
plt.grid(True)
plt.show()
```

 ### 13. **IMDb Rating vs Revenue**

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Rating",
    y="Revenue (Millions)"
)

plt.yscale("log")

plt.title("IMDb Rating vs Revenue")
plt.xlabel("IMDb Rating")
plt.ylabel("Revenue (Log Scale)")
plt.grid(True)
plt.show()
#10^-2 = 0.01 million
#10^-1 = 0.1  million 
#10^1 = 10   million
#10^2 = 100   million
#10^3 = 1000  million
```
